```python
import zipfile
import os

# Find the uploaded ZIP file
zip_files = [f for f in os.listdir('/content') if f.lower().endswith('.zip')]

print("ZIP files found:", zip_files)

if len(zip_files) != 1:
    raise ValueError(
            f"Expected exactly one ZIP file in /content, found {len(zip_files)}. "
                    "Check the filenames printed above."
                        )

zip_path = os.path.join('/content', zip_files[0])
extract_folder = '/content/global_warming_dataset'

                        # Extract the ZIP file
os.makedirs(extract_folder, exist_ok=True)

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
                            zip_ref.extractall(extract_folder)

print("ZIP extracted successfully!")
print("Extracted location:", extract_folder)
```

```python
csv_files = []

for root, dirs, files_in_folder in os.walk(extract_folder):
    for filename in files_in_folder:
            if filename.lower().endswith('.csv'):
                        csv_files.append(os.path.join(root, filename))

print("CSV files found:", len(csv_files))

for file_path in csv_files:
    print(file_path)
```

```python
import pandas as pd
import os

csv_path = csv_files[0]

df = pd.read_csv(csv_path)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Column": df.columns,
        "Data_Type": df.dtypes.astype(str).values,
            "Missing_Count": missing_values.values,
                "Missing_Percentage": (
                        missing_values.values / len(df) * 100
                            ).round(2)
                            })

display(missing_report)

print("Total missing cells:", df.isnull().sum().sum())
```

```python
# Count completely duplicated rows
duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
print("Dataset shape:", df.shape)
```

```python
# Complete missing-value report
missing_report = pd.DataFrame({
    "Data_Type": df.dtypes.astype(str),
        "Missing_Count": df.isnull().sum(),
            "Missing_Percentage": (
                    df.isnull().sum() / len(df) * 100
                        ).round(2)
                        })

print("Dataset shape:", df.shape)
print("Total missing cells:", df.isnull().sum().sum())
print("\nColumns containing missing values:")

display(missing_report[missing_report["Missing_Count"] > 0])
```

```python
print(
      "Duplicate Country-Year combinations:",
          df.duplicated(subset=["Country", "Year"]).sum()
          )

print("Unique countries:", df["Country"].nunique())
print("Minimum year:", df["Year"].min())
print("Maximum year:", df["Year"].max())
```

```python
display(df.describe().T.round(2))
```

```python
# Count how many times each Country-Year combination occurs
country_year_counts = (
    df.groupby(["Country", "Year"])
          .size()
                .reset_index(name="Record_Count")
                )

                # Show summary of the counts
print("Total records:", len(df))
print("Unique Country-Year combinations:", len(country_year_counts))
print(
                    "Country-Year combinations appearing more than once:",
                        (country_year_counts["Record_Count"] > 1).sum()
                        )

print("\nDistribution of records per Country-Year combination:")
display(country_year_counts["Record_Count"].value_counts().sort_index())

print("\nExamples of repeated Country-Year combinations:")
display(
                            country_year_counts[
                                    country_year_counts["Record_Count"] > 1
                                        ].head(10)
                                        )
```

```python
# Display the original records for repeated Country-Year combinations
repeated_keys = country_year_counts.loc[
    country_year_counts["Record_Count"] > 1,
        ["Country", "Year"]
        ]

repeated_records = df.merge(
            repeated_keys,
                on=["Country", "Year"],
                    how="inner"
                    )

display(repeated_records.sort_values(["Country", "Year"]).head(20))
```
