```python
missing_columns = [
      "Energy_Consumption_Per_Capita",
          "Fossil_Fuel_Usage",
              "Policy_Score",
                  "Average_Temperature",
                      "Ocean_Acidification",
                          "Biodiversity_Index"
                          ]

for column in missing_columns:
    median_value = df[column].median()
    df[column] = df[column].fillna(median_value)
print("Missing values filled successfully!")
```

```python
numeric_columns = df.select_dtypes(include="number").columns

summary = df[numeric_columns].describe().T

display(
    summary[
            ["min", "max", "mean", "50%", "std"]
                ]
                )
```

```python
numeric_columns = df.select_dtypes(include="number").columns

outlier_summary = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = (
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ).sum()

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outliers": outliers,
        "Outlier %": (outliers / len(df)) * 100
    })

# Build and display the summary once, after every column has been checked
outlier_table = pd.DataFrame(outlier_summary)

display(
    outlier_table.sort_values(
        "Outliers",
        ascending=False
    )
)
```

```python
outliers_only = outlier_table[
      outlier_table["Outliers"] > 0
      ].sort_values(
          "Outliers",
              ascending=False
              )


```

```python
print("Negative values check:")
print()

checks = {
    "Population": (df["Population"] < 0).sum(),
        "CO2_Emissions": (df["CO2_Emissions"] < 0).sum(),
            "Forest_Area": (df["Forest_Area"] < 0).sum(),
                "GDP": (df["GDP"] < 0).sum(),
                    "Renewable_Energy_Usage": (df["Renewable_Energy_Usage"] < 0).sum(),
                        "Methane_Emissions": (df["Methane_Emissions"] < 0).sum(),
                            "Average_Rainfall": (df["Average_Rainfall"] < 0).sum(),
                                "Solar_Energy_Potential": (df["Solar_Energy_Potential"] < 0).sum(),
                                    "Waste_Management": (df["Waste_Management"] < 0).sum(),
                                        "Per_Capita_Emissions": (df["Per_Capita_Emissions"] < 0).sum(),
                                            "Industrial_Activity": (df["Industrial_Activity"] < 0).sum(),
                                                "Air_Pollution_Index": (df["Air_Pollution_Index"] < 0).sum(),
                                                    "Biodiversity_Index": (df["Biodiversity_Index"] < 0).sum(),
                                                        "Fossil_Fuel_Usage": (df["Fossil_Fuel_Usage"] < 0).sum(),
                                                            "Energy_Consumption_Per_Capita": (df["Energy_Consumption_Per_Capita"] < 0).sum(),
                                                                "Policy_Score": (df["Policy_Score"] < 0).sum()
                                                                }

for column, count in checks.items():
  print(f"{column}: {count}")
```

```python
percentage_columns = [
      "Forest_Area",
          "Renewable_Energy_Usage",
              "Urbanization",
                  "Waste_Management",
                      "Industrial_Activity",
                          "Biodiversity_Index",
                              "Fossil_Fuel_Usage",
                                  "Policy_Score"
                                  ]
for column in percentage_columns:
  below_zero = (df[column] < 0).sum()
  above_hundred = (df[column] > 100).sum()

  print(f"\n{column}")
  print("Below 0:", below_zero)
  print("Above 100:", above_hundred)
```
